```python
libra = {}

def add_book(library, book_id, title, quantity):
    if book_id in library:
        library[book_id]['quantidade'] += quantity
        print(f"Livro '{title}' (ID: {book_id}) atualizado. Quantidade total: {library[book_id]['quantidade']}.")
    else:
        library[book_id] = {'titulo': title, 'quantidade': quantity}
        print(f"Livro '{title}' (ID: {book_id}) com {quantity} unidades adicionado.")

def lend_book(library, book_id):
    if book_id in library:
        if library[book_id]['quantidade'] > 0:
            library[book_id]['quantidade'] -= 1
            print(f"Empréstimo de '{library[book_id]['titulo']}' (ID: {book_id}) realizado com sucesso.")
            print(f"Quantidade restante: {library[book_id]['quantidade']}.")
        else:
            print(f"Livro '{library[book_id]['titulo']}' (ID: {book_id}) indisponível para empréstimo.")
    else:
        print(f"Livro com ID {book_id} não encontrado no sistema.")

def return_book(library, book_id):
    if book_id in library:
        library[book_id]['quantidade'] += 1
        print(f"Devolução de '{library[book_id]['titulo']}' (ID: {book_id}) registrada com sucesso.")
        print(f"Quantidade total: {library[book_id]['quantidade']}.")
    else:
        print(f"Livro com ID {book_id} não pertence ao nosso acervo ou não foi encontrado.")

def list_books(library):
    if not library:
        print("A biblioteca está vazia.")
    else:
        print("\n--- Livros na Biblioteca ---")
        for book_id, details in library.items():
            print(f"ID: {book_id}, Título: '{details['titulo']}', Quantidade: {details['quantidade']}")
        print("----------------------------")

cont = True
while cont:
    print("\n--- Boas vindas à biblioteca leitura ---")
    print("Operações: ")
    print("1 = Adicionar livro")
    print("2 = Emprestar livro")
    print("3 = Devolver livro")
    print("4 = Listar livros")
    print("5 = Sair")

    try:
        op = int(input("Qual operação você deseja? "))
    except ValueError:
        print("Entrada inválida. Por favor, digite um número de 1 a 5.")
        continue

    if op == 1:
        try:
            book_id = int(input("Qual o ID do livro? "))
            title = input("Qual o título do livro? ")
            quantity = int(input("Qual a quantidade a adicionar? "))
            if quantity < 0:
                print("A quantidade não pode ser negativa.")
                continue
            add_book(libra, book_id, title, quantity)
        except ValueError:
            print("Entrada inválida para ID ou quantidade. Por favor, digite números inteiros.")
    elif op == 2:
        try:
            book_id = int(input("Qual o ID do livro a emprestar? "))
            lend_book(libra, book_id)
        except ValueError:
            print("Entrada inválida para ID. Por favor, digite um número inteiro.")
    elif op == 3:
        try:
            book_id = int(input("Qual o ID do livro a devolver? "))
            return_book(libra, book_id)
        except ValueError:
            print("Entrada inválida para ID. Por favor, digite um número inteiro.")
    elif op == 4:
        list_books(libra)
    elif op == 5:
        cont = False
        print("Obrigado por usar a biblioteca!")
    else:
        print("Operação inválida. Por favor, escolha 1, 2, 3, 4 ou 5.")

    if cont:
        continue_choice = input("Deseja realizar outra operação? (s/n): ").lower()
        if continue_choice != 's':
            cont = False
            print("Obrigado por usar a biblioteca!")
```
